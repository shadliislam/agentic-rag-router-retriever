```python
from crewai import BaseLLM

from typing import Any, Dict, List, Optional, Union

import requests

class CourseAzureLLM(BaseLLM):

    def __init__(

        self,

        model: str,

        api_key: str,

        endpoint: str

    ):

        super().__init__(

            model=model,

            temperature=None

        )

        self.api_key = api_key

        self.endpoint = endpoint

    def call(

        self,

        messages: Union[str, List[Dict[str, str]]],

        tools: Optional[List[dict]] = None,

        callbacks: Optional[List[Any]] = None,

        available_functions: Optional[Dict[str, Any]] = None,

        **kwargs

    ) -> str:

        if isinstance(messages, str):

            messages = [

                {

                    "role": "user",

                    "content": messages

                }

            ]

        payload = {

            "messages": messages

        }

        headers = {

            "api-key": self.api_key,

            "Content-Type": "application/json"

        }

        response = requests.post(

            self.endpoint,

            headers=headers,

            json=payload,

            timeout=120

        )

        if not response.ok:

            raise RuntimeError(

                f"Course API request failed: "

                f"{response.status_code} {response.text}"

            )

        result = response.json()

        return result["choices"][0]["message"]["content"]

    def supports_function_calling(self) -> bool:

        return False

    def get_context_window_size(self) -> int:

        return 128000

print("Custom course LLM class created!")

 
```

```python
from datetime import datetime

from crewai import Agent, Task, Crew, Process

def clean_route(route_output):

    """

    Convert the CrewAI output into either PDF or WEB.

    """

    route_text = str(route_output).strip().upper()

    if "WEB" in route_text and "PDF" not in route_text:

        return "WEB"

    if "PDF" in route_text:

        return "PDF"

    raise ValueError(

        f"Router returned an unsupported route: {route_text}"

    )

async def run_agentic_rag(question):

    """

    Route a question to PDF or web search, retrieve evidence,

    and use the Retriever Agent to generate a grounded answer.

    """

    start_time = datetime.now()

    print("=" * 70)

    print("USER QUESTION:")

    print(question)

    print("=" * 70)

    # ---------------------------------------------------

    # STEP 1: ROUTER AGENT

    # ---------------------------------------------------

    router_task = Task(

        description=f"""

Analyze the following user question:

{question}

Choose exactly one retrieval route:

PDF:

Use when the question can be answered from the supplied

Transformer research paper. This includes questions about

attention, multi-head attention, positional encoding,

encoder-decoder architecture, training, or experimental results.

WEB:

Use when the question asks for current, latest, recent,

changing, or external information.

Return only one word: PDF or WEB.

""",

        expected_output="Exactly one word: PDF or WEB",

        agent=router_agent

    )

    router_crew = Crew(

        agents=[router_agent],

        tasks=[router_task],

        process=Process.sequential,

        verbose=True

    )

    raw_route = await router_crew.akickoff()

    route = clean_route(raw_route)

    print("\nROUTER DECISION:")

    print(route)

    # ---------------------------------------------------

    # STEP 2: RETRIEVAL

    # ---------------------------------------------------

    if route == "PDF":

        tool_name = "PDFSearchTool"

        source_name = "Transformer research paper"

        evidence = pdf_tool.run(

            query=question

        )

    else:

        tool_name = "TavilySearchResults"

        source_name = "Web search"

        evidence = web_tool.invoke(

            {"query": question}

        )

    evidence_text = str(evidence)

    # Prevent an excessively long prompt

    if len(evidence_text) > 12000:

        evidence_text = evidence_text[:12000]

    print("\nRETRIEVAL TOOL:")

    print(tool_name)

    print("\nSOURCE:")

    print(source_name)

    # ---------------------------------------------------

    # STEP 3: RETRIEVER AGENT

    # ---------------------------------------------------

    retriever_task = Task(

        description=f"""

Answer the user's question using ONLY the retrieved evidence.

Question:

{question}

Selected route:

{route}

Retrieved evidence:

{evidence_text}

Requirements:

1. Give a clear and concise answer.

2. Use only the supplied evidence.

3. Do not invent unsupported information.

4. If the evidence is insufficient, clearly say so.

5. End with: Source: {source_name}

""",

        expected_output=(

            "A grounded answer based only on the supplied evidence, "

            "followed by the source."

        ),

        agent=retriever_agent

    )

    retriever_crew = Crew(

        agents=[retriever_agent],

        tasks=[retriever_task],

        process=Process.sequential,

        verbose=True

    )

    answer = await retriever_crew.akickoff()

    duration = round(

        (datetime.now() - start_time).total_seconds(),

        2

    )

    # ---------------------------------------------------

    # STEP 4: OPERATIONAL TRACE

    # ---------------------------------------------------

    trace = {

        "timestamp": start_time.isoformat(timespec="seconds"),

        "question": question,

        "router_decision": route,

        "retrieval_tool": tool_name,

        "source": source_name,

        "answer": str(answer),

        "duration_seconds": duration

    }

    print("\n" + "=" * 70)

    print("FINAL ANSWER:")

    print(answer)

    print("\nOPERATIONAL TRACE:")

    print("Timestamp:", trace["timestamp"])

    print("Route:", trace["router_decision"])

    print("Tool:", trace["retrieval_tool"])

    print("Source:", trace["source"])

    print("Duration:", trace["duration_seconds"], "seconds")

    print("=" * 70)

    return trace

print("Combined Agentic RAG workflow created!")
```

```python
pdf_trace = await run_agentic_rag(

    "What is scaled dot-product attention?"

)
```

```python
web_trace = await run_agentic_rag(

    "What are the latest developments in Transformer architectures?"

)
```
